```python
# 毎回、最初に一度だけ実行してください
!pip install -U torch torchvision torchaudio torchdata torchtext transformers sentencepiece accelerate

import torch
import transformers
from transformers import AutoModelForCausalLM, AutoTokenizer
import ipywidgets as widgets
from IPython.display import display, clear_output
import re

model_name = "watashiha/watashiha-gpt-6b"
tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=False)
model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype=torch.bfloat16, low_cpu_mem_usage=True)

if torch.cuda.is_available():
    model = model.to("cuda")

# スタイル設定
text_style = {'description_width': 'initial'}  # descriptionの幅を調整
text_layout = widgets.Layout(width='80%', height='30px', margin='0px')
button_layout = widgets.Layout(width='80%', height='30px', margin='20px')

# テキストボックスのウィジェット
text = widgets.Text(
    value='',
    placeholder='ホラー映画の「○○○から逃げろ！」',
    description='お題:',
    disabled=False,
    style=text_style,
    layout=text_layout
)

# ボタンのウィジェット
button_run = widgets.Button(
    description="出題",
    button_style='success', # 'success', 'info', 'warning', 'danger' or ''
    layout=button_layout
)

# 出力表示用のウィジェット
output = widgets.Output(layout={'border': '1px solid black', 'width': '100%'})

# ボタンクリック時のコールバック関数
def on_button_clicked(b):
    with output:
        # 出力をクリア
        clear_output()
        # ここでAIの処理を行う
        odai = text.value
        input_text = f"お題:{odai}<SEP>回答:"
        token_ids = tokenizer.encode(input_text, add_special_tokens=False, return_tensors="pt").to(model.device)

        output_ids = model.generate(
            token_ids,
            do_sample=True,
            max_new_tokens=32,
            top_p=0.9,
            top_k=50,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )
        result = re.sub(".*回答:", "", tokenizer.decode(output_ids.tolist()[0], skip_special_tokens=True))        # 結果を大きな文字で表示
        display(widgets.HTML(value=f"<h2>{result}</h2>"))

# コールバック関数のリンク
button_run.on_click(on_button_clicked)

# ウィジェットの表示
display(text, button_run, output)
```
